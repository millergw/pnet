```python
# testing SNR calculations to make sure they match my expectations.
import pandas as pd
import numpy as np

# Example datasets
df1 = pd.DataFrame({
    'feature1': [10, 20, 30],
    'feature2': [5, 15, 25]
})
df2 = pd.DataFrame({
    'feature1': [1, 1, 1],
    'feature2': [10, 20, 30]
})

# Data grouped by dataset
df_imps_by_dataset = {'germline only': df1, 'somatic only': df2}
display(df_imps_by_dataset)

# Calculate signal-to-noise ratio (SNR) for each dataset
snr_per_dataset = {}
for dataset, df_imps in df_imps_by_dataset.items():
    snr = df_imps.mean(axis=0) / (df_imps.std(axis=0) + 1e-9)
    snr_per_dataset[dataset] = snr

# Print results
for dataset, snr in snr_per_dataset.items():
    print(f"SNR for {dataset}:\n{snr}\n")
```
